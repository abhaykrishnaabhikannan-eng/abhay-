```python
# Q1
data = [12, 15, 14, 10, 100, 13, 14, 15, 16, 14]
def find_range(numbers):
    return min(numbers), max(numbers)
```

```python
# Q2
data = [12, 15, 14, 10, 100, 13, 14, 15, 16, 14]
def find_mean(numbers):
    return sum(numbers) / len(numbers)
```

```python
 #Q3
data = [12, 15, 14, 10, 100, 13, 14, 15, 16, 14]
def find_median(numbers):
    sorted_nums = sorted(numbers)
    n = len(sorted_nums)
    mid = n // 2
    if n % 2 == 0:
        median = (sorted_nums[mid - 1] + sorted_nums[mid]) / 2
    else:
        median = sorted_nums[mid]
    return median
```

```python
# Q4. Mode
data = [12, 15, 14, 10, 100, 13, 14, 15, 16, 14]
def find_mode(numbers):
    frequency = {}
    for n in numbers:
        if n in frequency:
            frequency[n] += 1
        else:
            frequency[n] = 1
    max_freq = 0
    for key in frequency:
        if frequency[key] > max_freq:
            max_freq = frequency[key]
    modes = []
    for key in frequency:
        if frequency[key] == max_freq:
            modes.append(key)
    if len(modes) == 1:
        return modes[0]
    else:
        return modes
```

```python
 #Q5. Percentile
def find_percentile(numbers, percentile):
    sorted_nums = sorted(numbers)
    n = len(sorted_nums)
    index = (percentile / 100) * (n - 1)
    lower = int(index)
    upper = lower + 1
    if upper >= n:
        return sorted_nums[lower]
    fraction = index - lower
    value = sorted_nums[lower] + fraction * (sorted_nums[upper] - sorted_nums[lower])
    return value
```

```python
#Q6
def find_lower_bound(numbers):
    q1 = find_percentile(numbers, 25)
    q3 = find_percentile(numbers, 75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    return lower_bound
```

```python
#Q7
def find_outliers(numbers):
    q1 = find_percentile(numbers, 25)
    q3 = find_percentile(numbers, 75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = []
    for n in numbers:
        if n < lower_bound or n > upper_bound:
            outliers.append(n)
    return outliers
```

```python
#Q8

def find_std_dev(numbers):
    mean = find_mean(numbers)
    n = len(numbers)

    total = 0
    for x in numbers:
        total += (x - mean) ** 2

    variance = total / (n - 1)
    std_dev = variance ** 0.5
    return std_dev
```

```python
#Q9
def find_z_scores(numbers):
    mean = find_mean(numbers)
    std = find_std_dev(numbers)
    z_scores = []
    for x in numbers:
        z = (x - mean) / std
        z_scores.append(z)
    return z_scores
```

```python
# Q10
def pearson_correlation(x, y):
    n = len(x)
    mean_x = find_mean(x)
    mean_y = find_mean(y)
    numerator = 0
    for i in range(n):
        numerator += (x[i] - mean_x) * (y[i] - mean_y)
    sum_x_sq = 0
    for i in range(n):
        sum_x_sq += (x[i] - mean_x) ** 2
    sum_y_sq = 0
    for i in range(n):
        sum_y_sq += (y[i] - mean_y) ** 2
    denominator = (sum_x_sq * sum_y_sq) ** 0.5
    correlation = numerator / denominator
    return correlation
```

```python
data = [12, 15, 14, 10, 100, 13, 14, 15, 16, 14]

print("Range:", find_range(data))
print("Mean:", find_mean(data))
print("Median:", find_median(data))
print("Mode:", find_mode(data))
print("90th Percentile:", find_percentile(data, 90))
print("Lower Bound:", find_lower_bound(data))
print("Outliers:", find_outliers(data))
print("Std Dev:", find_std_dev(data))
print("Z-scores:", find_z_scores(data))
```
